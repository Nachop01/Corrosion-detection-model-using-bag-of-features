### Evaluation

```python
def evaluate_model(y_pred, y_test, model, labels=('corrosion', 'no corrosion')):
    """
    Evaluate the performance of a classification model using confusion matrix, accuracy, recall, specificity, precision, and F1 score.
    """
    cm = confusion_matrix(y_test, y_pred, labels=labels)
    print(f'Confusion matrix: \n {cm}')
    accuracy = accuracy_score(y_test, y_pred)
    print(f'Accuracy: {accuracy}')
    recall = recall_score(y_test, y_pred, average='binary', pos_label=labels[0])
    recall = round(recall, 3)
    print(f'Recall: {recall}')
    specificity = recall_score(y_test, y_pred, average='binary', pos_label=labels[1])
    specificity = round(specificity, 3)
    print(f'Specificity: {specificity}')
    precision = precision_score(y_test, y_pred, average='binary', pos_label=labels[0])
    precision = round(precision, 3)
    print(f'Precision: {precision}')
    f1 = f1_score(y_test, y_pred, average='binary', pos_label=labels[0])
    f1 = round(f1, 3)
    print(f'F1: {f1}')
    df_results = pd.DataFrame(columns=['Model', 'Accuracy', 'Recall', 'Specificity', 'Precision', 'F1'])
    result_i = {'Model': model, 'Accuracy': accuracy, 'Recall': recall, 'Specificity': specificity, 'Precision': precision, 'F1': f1}
    df_results = pd.concat([df_results, pd.DataFrame([result_i])], ignore_index=True)
    return df_results


def means_results(df_results):
    """
    Calculate the mean of the recall, specificity, precision, and F1 score.
    """
    mean_recall = df_results['Recall'].mean()
    mean_specificity = df_results['Specificity'].mean()
    mean_precision = df_results['Precision'].mean()
    mean_f1 = df_results['F1'].mean()
    mean_accuracy = df_results['Accuracy'].mean()
    return mean_recall, mean_specificity, mean_precision, mean_f1, mean_accuracy

```
